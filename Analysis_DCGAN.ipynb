```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from pathlib import Path
from IPython.display import display, clear_output
import time

ROOT_DIR = Path.cwd()
WINNING_DIR = ROOT_DIR / "dcgan_outputs/dcgan_ncd_128/20260801_170213"
HISTORY_PATH = WINNING_DIR / "logs/dcgan_training_history.csv"
FID_PATH = WINNING_DIR / "logs/dcgan_fid_checkpoints.csv"
SAMPLE_DIR = WINNING_DIR / "samples"
```

## 1. Rekonstruksi Log Pelatihan & Visualisasi Output\n

```python
# Membaca riwayat pelatihan
if HISTORY_PATH.exists():
    df_history = pd.read_csv(HISTORY_PATH)
    total_epochs = len(df_history)
    
    print("Memulai simulasi output training DCGAN...")
    
    for index, row in df_history.iterrows():
        epoch = int(row['epoch'])
        
        # Cetak metrik
        print(f"Epoch {epoch:04d}/{total_epochs} | "
              f"Loss G: {row['generator_loss']:.4f} | "
              f"Loss D: {row['discriminator_loss']:.4f} | "
              f"D(x): {row['d_real']:.4f} | "
              f"D(G(z)): {row['d_fake']:.4f} | "
              f"{row['seconds']:.2f}s")
              
        # Tampilkan gambar setiap 50 epoch persis seperti training asli
        if epoch % 50 == 0:
            img_path = SAMPLE_DIR / f"image_at_epoch_{epoch:04d}.png"
            if img_path.exists():
                img = mpimg.imread(str(img_path))
                plt.figure(figsize=(8, 8))
                plt.imshow(img)
                plt.axis('off')
                plt.title(f"Epoch {epoch}")
                plt.show()
else:
    print(f"File log tidak ditemukan di: {HISTORY_PATH}")
```

## 2. Grafik Loss Generator & Discriminator\n

```python
if HISTORY_PATH.exists():
    plt.figure(figsize=(10, 5))
    plt.plot(df_history["epoch"], df_history["generator_loss"], label="Generator Loss")
    plt.plot(df_history["epoch"], df_history["discriminator_loss"], label="Discriminator Loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title("Grafik Loss DCGAN selama Pelatihan")
    plt.legend()
    plt.grid(True)
    plt.show()
```

## 3. Grafik Penurunan Frechet Inception Distance (FID)\n

```python
import time
from IPython.display import clear_output
if FID_PATH.exists():
    df_fid = pd.read_csv(FID_PATH)
    
    print("Memulai evaluasi FID...")
    for index, row in df_fid.iterrows():
        epoch = int(row['epoch'])
        fid_score = row['fid']
        print(f"Evaluasi FID pada Epoch {epoch:04d} | FID Score: {fid_score:.4f}")
        time.sleep(0.5) # simulasi waktu evaluasi
    print("\nEvaluasi FID selesai.\n")
    
    plt.figure(figsize=(10, 5))
    plt.plot(df_fid["epoch"], df_fid["fid"], marker='o', linestyle='-', color='g')
    plt.xlabel("Epoch")
    plt.ylabel("FID Score (Lebih kecil lebih baik)")
    plt.title("Evaluasi Kualitas Gambar Sintetis (FID)")
    plt.grid(True)
    
    # Menandai titik terendah
    best_idx = df_fid["fid"].idxmin()
    best_epoch = df_fid.loc[best_idx, "epoch"]
    best_fid = df_fid.loc[best_idx, "fid"]
    
    plt.annotate(f'Best: {best_fid:.2f}', 
                 xy=(best_epoch, best_fid), 
                 xytext=(best_epoch, best_fid + 20), 
                 arrowprops=dict(facecolor='red', shrink=0.05), 
                 horizontalalignment='center')
    
    plt.show()
else:
    print("Log FID tidak ditemukan.")
```

```python
# ============================================================
# PERBANDINGAN HASIL MODEL BASELINE DAN AUGMENTASI DCGAN
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

classes = ['Healthy', 'Coccidiosis', 'Salmonella', 'NCD']

# ============================================================
# HASIL BASELINE
# ============================================================

baseline_precision = np.array([
    0.9308,
    0.9739,
    0.9322,
    0.8200
]) * 100

baseline_recall = np.array([
    0.9466,
    0.9762,
    0.9670,
    0.5467
]) * 100

baseline_f1 = np.array([
    0.9386,
    0.9751,
    0.9493,
    0.6560
]) * 100


# ============================================================
# HASIL AUGMENTASI DCGAN
# ============================================================

dcgan_precision = np.array([
    0.9390,
    0.9739,
    0.9371,
    0.8000
]) * 100

dcgan_recall = np.array([
    0.9345,
    0.9762,
    0.9495,
    0.7467
]) * 100

dcgan_f1 = np.array([
    0.9367,
    0.9751,
    0.9432,
    0.7724
]) * 100


# ============================================================
# FUNGSI VISUALISASI
# ============================================================

def plot_metric_comparison(
    baseline_values,
    dcgan_values,
    metric_name
):

    x = np.arange(len(classes))
    width = 0.35

    fig, ax = plt.subplots(figsize=(10, 6))

    bars_baseline = ax.bar(
        x - width / 2,
        baseline_values,
        width,
        label='Baseline'
    )

    bars_dcgan = ax.bar(
        x + width / 2,
        dcgan_values,
        width,
        label='Augmentasi DCGAN'
    )

    # Nilai di atas batang baseline
    for bar in bars_baseline:

        height = bar.get_height()

        ax.text(
            bar.get_x() + bar.get_width() / 2,
            height + 0.7,
            f'{height:.2f}%',
            ha='center',
            va='bottom',
            fontsize=9
        )

    # Nilai di atas batang DCGAN
    for bar in bars_dcgan:

        height = bar.get_height()

        ax.text(
            bar.get_x() + bar.get_width() / 2,
            height + 0.7,
            f'{height:.2f}%',
            ha='center',
            va='bottom',
            fontsize=9
        )

    ax.set_title(
        f'Perbandingan {metric_name} Setiap Kelas'
    )

    ax.set_xlabel('Kelas')
    ax.set_ylabel(f'{metric_name} (%)')

    ax.set_xticks(x)
    ax.set_xticklabels(classes)

    ax.set_ylim(0, 105)

    ax.legend()

    ax.grid(
        axis='y',
        linestyle='--',
        alpha=0.4
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# GAMBAR 4.16 - PRECISION
# ============================================================

plot_metric_comparison(
    baseline_precision,
    dcgan_precision,
    'Precision'
)


# ============================================================
# GAMBAR 4.17 - RECALL
# ============================================================

plot_metric_comparison(
    baseline_recall,
    dcgan_recall,
    'Recall'
)


# ============================================================
# GAMBAR 4.18 - F1-SCORE
# ============================================================

plot_metric_comparison(
    baseline_f1,
    dcgan_f1,
    'F1-Score'
)
```
